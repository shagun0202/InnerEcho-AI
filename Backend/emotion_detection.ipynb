```python
from transformers import pipeline
```

```python
emotion_classifier = pipeline(
    task="text-classification",
    model="j-hartmann/emotion-english-distilroberta-base",
    top_k=None
)

print("✅ Emotion Detection Model Loaded Successfully")
```

```python
def predict_emotion(text):

    prediction = emotion_classifier(text)

    scores = prediction[0]

    top_prediction = max(scores, key=lambda x: x["score"])

    return {
        "input_text": text,
        "emotion": top_prediction["label"],
        "confidence": round(top_prediction["score"] * 100, 2),
        "all_predictions": scores
    }
```

```python
user_text = input("Enter your thoughts: ")

result = predict_emotion(user_text)

print("\n========== Prediction ==========")
print("Input :", result["input_text"])
print("Emotion :", result["emotion"])
print("Confidence :", result["confidence"], "%")
```

```python
examples = [

    "I am very happy today.",

    "I feel lonely.",

    "I am scared of tomorrow.",

    "Nobody understands me.",

    "I am excited for my vacation.",

    "I am angry with my friend.",

    "I failed my exam.",

    "I got selected for an internship."

]

for sentence in examples:

    result = predict_emotion(sentence)

    print("="*60)
    print("Input :", result["input_text"])
    print("Emotion :", result["emotion"])
    print("Confidence :", result["confidence"], "%")
```

```python
text = "I am feeling stressed because of exams."

result = predict_emotion(text)

print("All Emotion Scores\n")

for item in result["all_predictions"]:

    print(f"{item['label']} : {round(item['score']*100,2)} %")
```
